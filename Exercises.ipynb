# Write a Python program to calculate body mass index

```python
weight = float(input("Enter your weight in kg: "))
height = float(input("Enter your height in m: "))

BMI = round(weight / (height ** 2), 2)

print("BMI: ", BMI)
```

```python
x = float(input("Input number 1: "))
y = float(input("Input number 2: "))

answer = x ** 2 + 2 * x * y + y ** 2

print("({} + {}) ^ 2 = {})".format(x, y, answer))
```

# Write a Python program that accepts an integer (n) and computes the value of
# n+nn+nnn. Sample value of n is 5 Expected Result : 615

```python
a = int(input("Number: "))

n = int("%s" % a)
nn = int("%s%s" % (a, a))
nnn = int("%s%s%s" % (a, a, a))

print(n + nn + nnn)
```

# Write a Python program to calculate the sum of the digits in an integer (of 4 digits).

```python
num = int(input("Enter a 4 digit number: "))

x = (num // 1000)
x1 = (num - x*1000) // 100
x2 = (num - x1*100) // 10
x3 = (num - x2*10) 

print(x + x1 + x2 + x3)


```

```python
num = int(input("Enter a 4 digit number: "))

x = (num // 1000)
x1 = (num - x*1000) // 100
x2 = (num - x*1000 - x1*100) // 10
x3 = (num - x*1000 - x1*100 - x2*10)

print(x+x1+x2+x3)
```
