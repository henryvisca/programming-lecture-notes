```python
turn = input("Turn left or right?").strip().lower()  ## correcting wrong inputs like upper cases
while turn != "left":
    turn = input("Turn left or right?")
print("Route confirmed")
```

```python
turn = input("Turn left or right?").strip() ## eliminating spaces
while turn != "left":
    turn = input("Turn left or right?")
print("Route confirmed")
```
